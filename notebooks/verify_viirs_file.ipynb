# Verify VIIRS File
This notebook will verify the validity of the `VIIRS_1992.tif` file using Python libraries such as `rasterio` and `matplotlib`.

```python
# Import required libraries
import rasterio
import matplotlib.pyplot as plt

# Define the file path
file_path = "data/processed/ntl/VIIRS_1992.tif"

# Open the file and check its validity
try:
    with rasterio.open(file_path) as src:
        print("File successfully opened.")
        print("Metadata:", src.meta)

        # Read the first band
        band1 = src.read(1)

        # Plot the data
        plt.figure(figsize=(10, 6))
        plt.imshow(band1, cmap='gray')
        plt.colorbar(label='Pixel Values')
        plt.title("VIIRS 1992 - Band 1")
        plt.show()
except Exception as e:
    print("Error opening file:", e)
```
