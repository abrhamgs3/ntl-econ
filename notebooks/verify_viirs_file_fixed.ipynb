```python
# Debugging: Check if the file exists and its format
import os
from rasterio.errors import RasterioIOError

# Define the file path
file_path = "data/vector/VIIRS_1992.tif"

# Verify if the file exists
if not os.path.exists(file_path):
    print(f"File not found: {file_path}")
else:
    print(f"File found: {file_path}")

    # Try opening the file with rasterio
    try:
        with rasterio.open(file_path) as src:
            print("File successfully opened.")
            print("Metadata:", src.meta)
    except RasterioIOError as rio_err:
        print("RasterioIOError:", rio_err)
    except Exception as e:
        print("An unexpected error occurred:", e)
```

```python
# Reprocess the file using gdal_translate to ensure it is a valid GeoTIFF
import subprocess
import os

# Define the output file path
output_file_path = "data/processed/ntl/VIIRS_1992_fixed.tif"

# Check if the file exists
if os.path.exists(file_path):
    try:
        # Run gdal_translate to convert the file to a valid GeoTIFF
        command = [
            "gdal_translate",
            "-of", "GTiff",
            file_path,
            output_file_path
        ]
        subprocess.run(command, check=True)
        print(f"File successfully reprocessed and saved to: {output_file_path}")
    except subprocess.CalledProcessError as e:
        print("Error during file reprocessing:", e)
else:
    print(f"Input file does not exist: {file_path}")
```
